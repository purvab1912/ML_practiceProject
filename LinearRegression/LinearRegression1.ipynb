```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# 1. Create some simple data
np.random.seed(0)
X = 2 * np.random.rand(100, 1)  # 100 values between 0 and 2
y = 4 + 3 * X + np.random.randn(100, 1)  # true line: y = 4 + 3x + noise
```

```python
# 2. Fit Linear Regression
model = LinearRegression()
model.fit(X, y)
```

```python
# 3. Predict using the model
X_new = np.array([[0], [2]])
y_pred = model.predict(X_new)
```

```python
# 4. Plotting
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', label='Data points')
plt.plot(X_new, y_pred, color='red', linewidth=2, label='Best-fit line')
plt.title('Linear Regression: Best-Fit Line')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()

# 5. Print the learned parameters
print(f"Intercept (b): {model.intercept_[0]:.2f}")
print(f"Slope (w): {model.coef_[0][0]:.2f}")
```
